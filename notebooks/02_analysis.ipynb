## Gráficos

```python
plt.hist(df_pokemon['hp'], bins=20)
plt.title('Distribuição de HP dos Pokémon')
plt.xlabel('HP')
plt.ylabel('Quantidade')
plt.savefig('../images/distribuicao_hp.png')
plt.show()
```

```python
medias = df_pokemon.groupby('type1')['attack'].mean().sort_values(ascending=False)
```

```python
plt.bar(medias.index, medias.values)
plt.title('Média de Attack por Tipo')
plt.xlabel('Tipo')
plt.ylabel('Attack médio')
plt.xticks(rotation=90)
plt.savefig('../images/attack_tipo.png')
plt.show()
```

```python
cores = df_pokemon['is_legendary'].map({0: 'blue', 1: 'red'})
plt.scatter(df_pokemon['attack'], df_pokemon['defense'], c=cores)
plt.title('Attack x Defense')
plt.xlabel('Attack')
plt.ylabel('Defense')
plt.savefig('../images/attack_defense.png')
plt.show
```

```python
geracoes = [df_pokemon[df_pokemon['generation'] == g]['speed'].values for g in sorted(df_pokemon['generation'].unique())]

plt.boxplot(geracoes)
plt.title('Speed por Geração')
plt.xlabel('Geração')
plt.ylabel('Speed')
plt.savefig('../images/speed_geracao.png')
plt.show()
```

```python
media_geracao = df_pokemon.groupby('generation')['total_power'].mean()

plt.plot(media_geracao.index, media_geracao.values)
plt.title('Poder Total Médio por Geração')
plt.xlabel('Geração')
plt.ylabel('Poder Total Médio')
plt.savefig('../images/media_total_geracao.png')
plt.show()
```
